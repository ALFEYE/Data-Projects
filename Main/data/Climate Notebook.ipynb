```python
import pandas as pd

df = pd.read_csv("en_climate_daily_MB_5023262_2015_P1D.csv")
```

```python
df = df[['Date/Time',
         'Total Precip (mm)', 
         'Mean Temp (°C)', 
         'Spd of Max Gust (km/h)']]

df = df.dropna()


#Low/No precipitation days
high_precip = df[df['Total Precip (mm)'] > 10]

#High precipitation days
low_precip = df[df['Total Precip (mm)'] == 0]


#Compare Avg
print("> High Precipitation Days <")
print(high_precip.mean(numeric_only = True))

print("> Low/No Precipitation Days <")
print(low_precip.mean(numeric_only = True))
```

```python
df.to_csv('2015 - 2025 Weather.csv', index = False)
```

```python
df[['Total Precip (mm)', 'Mean Temp (°C)']].corr()
```

```python
import matplotlib.pyplot as plt

labels = ['High Precip', 'Low/No Precip']
temps = [high_precip['Mean Temp (°C)'].mean(),
         low_precip['Mean Temp (°C)'].mean()]


```

```python

plt.figure(figsize=(6,4))

plt.bar(labels, temps, color=['blue', 'red'],)

plt.title("Average Temperature: Precipitation Comparison")
plt.ylabel("Temperature (°C)")
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.show()
```

```python


plt.figure(figsize=(7,5))

plt.scatter(
    df['Total Precip (mm)'],
    df['Mean Temp (°C)'],
    alpha=0.5, 
    color = 'green'
)

plt.title("Precipitation vs Temperature Relationship")
plt.xlabel("Precipitation (mm)")
plt.ylabel("Temperature (°C)")
plt.grid(True, linestyle='-.', alpha=0.5)

plt.show()
```
