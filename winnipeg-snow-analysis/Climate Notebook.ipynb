```python
import pandas as pd

df = pd.read_csv("en_climate_daily_MB_5023262_2015_P1D.csv")
```

```python
df = df[['Date/Time',
         'Total Precip (mm)', 
         'Mean Temp (°C)', 
         'Spd of Max Gust (km/h)']]

df = df.dropna()


#Low/No precipitation days
high_precip = df[df['Total Precip (mm)'] > 10]

#High precipitation days
low_precip = df[df['Total Precip (mm)'] == 0]


#Compare Avg
print("> High Precipitation Days <")
print(high_precip.mean(numeric_only = True))

print("> Low/No Precipitation Days <")
print(low_precip.mean(numeric_only = True))
```

```python
import matplotlib.pyplot as plt

labels = ['High Precip', 'Low/No Precip']
temps = [high_precip['Mean Temp (°C)'].mean(),
         low_precip['Mean Temp (°C)'].mean()]

```

```python
plt.bar(labels, temps)
plt.title("Average Temperature Contrast")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
plt.scatter(df['Total Precip (mm)'], df['Mean Temp (°C)'])
plt.xlabel("precipation (mm)")
plt.ylabel("Temperature (°C)")
plt.title("Precipitation vs Temperature")
plt.show()
```
